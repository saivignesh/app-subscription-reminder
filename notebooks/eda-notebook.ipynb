# Set up

```python
from google.colab import userdata
from pathlib import Path
import kagglehub
import os

competition_name = 'house-price-prediction-challenge-regression'
kaggle_path = Path(f'/kaggle/input/competitions/{competition_name}')
colab_path  = Path(f'/content/input/competitions/{competition_name}')
data_path = kaggle_path if kaggle_path.exists() else colab_path
if not data_path.exists():
    os.environ['KAGGLE_API_TOKEN'] = userdata.get('KAGGLE_API_TOKEN')
    kagglehub.competition_download(competition_name,output_dir=data_path)
```
